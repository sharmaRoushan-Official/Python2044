### 4. Types of Variables in Python

Python automatically identifies data types.

A. Integer(int)

B. Float(float)

C. String(str)

```python
# A. Integer(int)
age = 30
print(age,type(age))

# B. Float(float)

price = 99.99
print(price,type(price))


# C. String(str)
city = "NOIDA"
print(city, type(city))

# D. Boolean(bool)
is_active = True
print(is_active,type(is_active))
```
