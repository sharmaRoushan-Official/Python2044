##### 4. Calculate the sum of numbers from 1 to n

```python
# for loop -

n = int(input("Enter value :"))

total = 0

for i in range(1,n+1):
    total = total+i

print(f"sum of 1 to {n} =", total)
```

##### 6. Print all numbers divisible by 3 and 5 from 1 to 100.

```python
# for loop -

for i in range(1,100):
    if i%3 != 0 or i%5 != 0:
        continue
    print(i)
```

```python
# while loop -

i = 1

while i <= 100:
  if i%3 == 0 and i%5 == 0:
    print(i)
  i = i+1
```

##### 14. Find the sum of digits of a number using a while loop.

```python
num = int(input("Enter value: "))

digits = []
    # Empty list data set used to saperate each digit from value

while num > 0:
    digits.append(num%10)
    # append used to etract last digit from value --> ex... 12345 >> 1234 > 5, 123 >> 4, 12 >> 3

    num //= 10
    # //= 10, used to remove last digit from value with every iteration --> 12345 >> 1234 >> 123 >> 12 >> 1 >> 0

digits.reverse()
    # append starts extract last digit from value, in looping last digit come to first position while first digit goes to last position so that reverse function used to arrange digits in assending order.

total = sum(digits)
    # sum funtion used to make sum of digits present in value.

print("sum of digits in value: ",total)
```

##### 18. Print all prime numbers between 1 and 100.

```python
# Soln -


for i in range(1,101):
    # i represents the number being checked.
    count = 0
    # count records how many divisors divide evenly into i.

    for j in range(1,i+1):
        # j checks every possible divisor from 1 to i.
        if i%j == 0:
            count = count + 1

    if count == 2:
        # if count == 2 confirms that the number is prime.
        print(i)
```

##### 20. Print Floyd's Triangle:



```python
# Soln-

# There are two for loops:
# Outer loop (i) controls the number of rows.
# Inner loop (j) controls how many numbers are printed in each row.
# num = num + 1 increases the number after printing it.
# print() moves the cursor to the next line after each row.

num = 1
n = 5

for i in range(1,n):
    for j in range(1,i+1):
        print(num,end=" ")
        num = num + 1

    print()
```

##### 22. Calculate the average of numbers in a list.

```python
list = [1,2,3,4,5,6]

total = 0

for i in list:
    total = total + i

count = len(list)

avg = total/count

print(f"Sum of {list} : ",total)
print(f"Count of {list} : ", count)
print(f"Average of {list} : ", avg)
```
