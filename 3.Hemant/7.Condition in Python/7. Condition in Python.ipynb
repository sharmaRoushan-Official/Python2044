### 4. lader if : if - elif - else
----
#### An if-elif-else statement checks multiple conditions one by one. The program executes the block of code under the first condition that is true. If none of the conditions are true, the code block under the else statement is executed.
----

#### How It Works

1. If Condition:  The program first evaluates the condition following the if
2. Elif Condition:  If the if condition is false, the program evaluates the condition following the elif You can have multiple elif conditions.
3. Else Condition:  If none of the if or elif conditions are true, the program executes the code block under the else

----

```python
# Example :

childAge = int(input("Enter your baby age: "))

if childAge>5:
    print("Go to primary school")
elif childAge == 5:
    print("Go to kindergarten")
else:
    print("Baby")
```

### 6. Nested Statement
----

##### A "nested if" statement in Python is simply an if, if-else, or if-elif-else block placed inside another conditional block. This structure allows you to check further sub-conditions only after a primary condition evaluates to True


##### Syntax
if condition 1:
    
    if condition 2:

```python
# Example - voting eligibility

# Condition 1 - Voter age must be completed 18 years
# Condition 2 - Voter must be citizen of India

age = int(input("Enter age: "))
country = input("Enter country: ")

if age>=18:
    if country =="India":
        print("Elegible for voting")
    else:
        print("Not elegible for voting")
else:
    print("Not elegible for voting")
```

### 13. Case Statement

```python
# Example - Week has 7 days and those are sequence of 1 to 7

# program will show Day name by entering number value between 1 to 7

day = int(input("Enter day number: "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid input")
```

```python
# identify the charactor whether it upper letter / lower letter / digit / symbol

# number - 48 to 57
# symbols - 32 to 47 and 58 to 64 and 91 to 96 and 123 to 126
# upper letter - 65 to 90
# lower letter - 97 to 122

x = input("Enter any charactor: ")

if ord(x) >= 48 and ord(x) <= 57:
    print(f"{x} is Number")
    
elif ord(x) >= 65 and ord(x) <= 90:
    print(f"{x} is Upper letter")

elif ord(x) >= 97 and ord(x) <= 122:
    print(f"{x} is Lower letter")

elif 32 <= ord(x) <= 47:
    print(f"{x} is symbol")

elif 58 <= ord(x) <= 64:
    print(f"{x} is symbol")
    
elif 91 <= ord(x) <= 96:
    print(f"{x} is symbol")
    
elif 123 <= ord(x) <= 126:
    print(f"{x} is symbol")

else:
    print("Unknown charactor")
```
