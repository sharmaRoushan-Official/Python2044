# 3.A compay will decided to give bonus to the employee according to the following criteria:
# Time period of the service -------------------> Bonus
# More then 10 Years ---------------------------> 10%
# >=6 and <=10 ---------------------------------> 8%
# <6 -------------------------------------------> 5%
# Ask the user for their salary and years of service and then calcualte the bonus amount is, 
# and display the total salary after increment of bonus.


```python
serviceTime = int(input("Enter service time in the organization :"))

salary = int(input("Enter salary of employee"))

if serviceTime>10:
    salary=salary+(salary/10)
    print("Total salary after increment of bonus is",salary)

elif serviceTime>=6 and serviceTime<=10:
    salary=salary+(8/100*salary)
    print("Total salary after increment of bonus is",salary)

else:
    salary=salary+(5/100*salary)
    print("Total salary after increment of bonus is",salary)
```

```python
# 7. Accept the following from the user and calculate the percentage of the class attendence.
# Total number of working days.
# Total number of absent.
# if percentage is less than 75%, then student will not sit the exam.

day = int(input("Enter total working days of student :"))
absent = int(input("Enter total number of absentism of student :"))

present = (day-absent)

percentage = (present/day)*100

if percentage<75:
    print("Student not eligible to sit in the Exam")

else:
    print("Student eligible for the Exam")

```

```python
# 15.Write a Python program to check whether a character is a *digit, **uppercase letter, **lowercase letter, or **special character*.

char = input("Enter any thing as input :")

if char.isupper():
    print("Enter character is a uppercase letter")

elif char.islower():
    print("Enter character is a lowercase letter")

elif char.isdigit():
    print("Enter character is digit")

else:
    print("Enter character is special character")

```
