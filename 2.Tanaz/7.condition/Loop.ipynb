```python
# 1.Print numbers from 1 to 10 using a for loop.

for i in range(1,11):
    print(i)
```

```python
# 2.Print all even numbers between 1 and 100.

for i in range(1,101):
    if i % 2 == 0:
        print(f"{i} is even number")
    else:
        print(f"{i} is odd number")
```

```python
# 3.Print all numbers divisible by 3 and 5 from 1 to 100.

for i in range(1,101):
    if i % 3 == 0 and i % 5 == 0:
        print(f"{i} is divisible by both 3 and 5") #  We can also add else condition for more attractive
```

```python
# 4.Print the reverse of a given number (e.g., 123 → 321).

num = int(input("Enter any number that you want reverse :"))

k = 0

for i in range(num):
    if num == 0:
        break
    j = num % 10
    k = k * 10 + j
    num = num // 10
    
print("Reverse :",k)
```

```python
# 5.Print the table of a given number (e.g., input 5 → 5, 10, 15, … 50).

num = int(input("Enter a number for create table :"))

for i in range (1,11):
   print(f"{num}X{i}={num*i}")
```

```python
# 6.Calculate the sum of numbers from 1 to n.

n = int(input("Enter any number for calculation :"))

addition = 0

for i in range (1,n):
    print(i)
    addition = i + addition
print("Sum :", addition)
    
```
