**# 4.Print the reverse of a given number (e.g., 123 → 321).**

```python
# 4.Print the reverse of a given number (e.g., 123 → 321).

num = int(input("Enter any number that you want reverse :"))

k = 0

for i in range(num):
    if num == 0:
        break
    j = num % 10
    k = k * 10 + j
    num = num // 10
    
print("Reverse :",k)
```

```python
num = int(input("Enter any number that you want reverse :"))

i = 1

k = 0

while i<=num:
    i+=1
    j = num % 10
    k = k*10+j
    
    num=num//10
    
print("Reverse :",k)
```

**5.Calculate the factorial of a number using a loop.**

```python
val = int(input("Enter any value :"))

fact = 1

for i in range(1,val+1):

    fact = fact * i

print(fact)
    
```

```python
val = int(input("Enter any value :"))

i = 1
fact = 1

while i<=val:

    fact = fact * i
    i+=1

print(fact)
    
```

**# 10.Print only vowels from a string input by the user.using for & while loop**

```python
string = input("Enter any type of word :")

vowels = "aeiouAEIOU"

for char in string:
    if char in vowels:
        print(char)
```

16. Print this pattern:

```python
count = int(input("Enter how many row want to display pattern :"))

for i in range(count+1): # 1 , 1
    for j in range(i):   # 1 , 1
        print("*",end="") # * 
    print()
```
