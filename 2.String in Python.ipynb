### 2. Python String

1. What is a String in Python?

A `string` in Python is sequence of Unicode chracter enclosed in quotes.


```python
s1 = "Hello"  # double qoute
s2 = 'world' # single quote
s3 = '''Multi-line Comment'''  # triple quote

print(s1)
print(s2)
print(s3)
```

### 3. String Indexing (Address of Characters)

Each chararacter has a position


|Indexing|P|y|t|h|o|n|
|---|---|---|---|---|---|---|
|Positive|0|1|2|3|4|5|
|Negative|-6|-5|-4|-3|-2|-1|

```python
## Positive Indexing
x = "Python"
print(x[0])
print(x[1])
print(x[2])
print(x[3])
print(x[4])
print(x[5])
```

```python
### Negative Indexing
x = "Python"

print(x[-1])
print(x[-2])
print(x[-3])
print(x[-4])
print(x[-5])
print(x[-6])
```

```python
name = "HemanP"
print(name)

name[5] = "t"
print(name)
```

### important string method

1. upper()
2. lower()
3. capitalize()
4. title()
5. swapcase()

```python
s = "hELLo WorLD"  # random case
print("Orignal String:",s)

## upper() - all characters in upper case

s1 = s.upper()
print(s1)


## lower() 

s2 = s.lower()
print(s2)

## capitalize()
s3 = s.capitalize()
print(s3)

## title
s4 = s.title()
print(s4)


## swapcase()
s5 = s.swapcase()
print(s5)
```

### Searching and Finding 

```python
s = "Hello World"
print(s)


print(s.find("a"))  # -1
print(s.find("H"))
print(s.find("w"))  
print(s.find("o"))
print(s.find("or"))
```

```python
### Checking Method: Boolean Return - True/False

# isalpha()  - alphabets - 26
# isdigit() - "0-9"
# isalnum() - "abc123"
# islower() - "abc"
# isupper() - "ABC"
```

```python
x = "Hello world"
y = "12345"
z = "abc123"
a = "abc"
u = "HELLO"
print(x.isalpha())
print(y.isdigit())
print(z.isalnum())
print(x.isalnum())  # True
print(y.isalnum())
print(a.islower())
print(u.isupper())
print(x.islower())
```

## Triming: White space remove

* strip - remove white space from both side
* lstrip - remove white space from left side
* rstrip - rmeove white space from right side

```python
s = "          Hello World          "
print("Original String:",s)
print(s.strip())  # both side space remove
print(s.lstrip()) # left side space remove
print(s.rstrip()) # right side space remove
```
