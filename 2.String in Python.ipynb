### 2. Python String

1. What is a String in Python?

A `string` in Python is sequence of Unicode chracter enclosed in quotes.


```python
s1 = "Hello"  # double qoute
s2 = 'world' # single quote
s3 = '''Multi-line Comment'''  # triple quote

print(s1)
print(s2)
print(s3)
```

## KeyPonints:

* Strings are immutable(cannot be changed after creation)
* Indexed(Ordered collection)
* Supports Slicing, iteration, and many built-in methods

### 3. String Indexing (Address of Characters)

Each chararacter has a position


|Indexing|P|y|t|h|o|n|
|---|---|---|---|---|---|---|
|Positive|0|1|2|3|4|5|
|Negative|-6|-5|-4|-3|-2|-1|

```python
## Positive Indexing
x = "Python"
print(x[0])
print(x[1])
print(x[2])
print(x[3])
print(x[4])
print(x[5])
```

```python
### Negative Indexing
x = "Python"

print(x[-1])
print(x[-2])
print(x[-3])
print(x[-4])
print(x[-5])
print(x[-6])
```
