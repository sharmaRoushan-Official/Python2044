### OOPs in Python


### 1. What is OOPs?

* OOP is a programming paradigm[A programming paradigm is a fundamental style or "Way of thinking" about how to structure and write your code.] that organizes software design around objects, rather than function and logics.


### Why OOPs

* Code resuability
* Modularity - Modularity in OOPs means is the practice of breaking a complex software program into small and modular chunks, indenepent , and self-contained units of functions.
* Scalability - [how well an object-oriented codebase, design, or architecture handles growing complexity, larger data volumes, or increased functional requirements without losing performance or becoming unmaintainable]
* Easy maintaince
* Real-World Modeling

### 2. Class 

A `class` is a blueprint or template to create any object.

syntax:


```python
class Student:
    pass
```

```python
help(Student)
```

Key Points:

* Defined using `class` keyword
* Can Contains varaibles and methods
* Naming convention: PascalCase


```python
class Student:  # class is keyword Student is the name of class
    def display(self):  # def display is method 
        print("Welcome to Python OOPs")


tanaz = Student()
print(tanaz)
```

```python
tanaz.display()
```

```python
id(tanaz)
```

```python
x = Student()
x.display()
```
