### 1. Inheritance 

a). Single/Simple Inheritance

Meaning:

* In `Single/Simple` Inheritance, a child inherits from only one parent class.

### 📌 Structure

`Parent ➺ Child`


### Why it is used?
* Code Reusability
* Child class can use parent class property
* Easy to understand and maintain

### Real-Life Example

* Parent- Vehicle
* Child - Car

A Car is a Vehicle

````python
class Vehicle:  # parent
    def start(self):
        print("Vehicle Started")

class Car(Vehicle):  # child - derived child 
    def drive(self):
        print("Car is started") 


# obj = Vehicle()
# print(obj.start())

# obj1 = Car()
# print(obj1.drive())

obj1 = Car()
obj1.start()
obj1.drive()
````

Note:

Single Inheritance is most basic and safest inheritance. It avoids complexity and ambiguity.

### 2. Multiple Inheritance

Meaning:

In `Multiple Inheritance`, a child class inherits from more than one parents class.

#### 📌 Strcuture
```
     parent1          parent2
           \         /
            \       /
             \     /
              child1
```
#### Why it is used?
* When one class needs feature from multiple classes
* Helps to combine functionalities

### Real-Life Example
* Parent1 = Camera
* Parent2 = Telephone
* Parent3 = Internet
* Mobile = Smart Phone

### SmartPhone is both Phone and a Camera

````python
class Phone:  # parent1
    def call(self):
        print("Calling......")

class Camera:  # parent2
    def clickPic(self):
        print("Photo Clicked.....")

class SmartPhone(Camera,Phone):  # derived child 
    def internet(self):
        print("Google Connected")

obj1 = SmartPhone()
obj1.call()
obj1.clickPic()
obj1.internet()
````

````python
print(SmartPhone.mro())
````

### 3. Multilevel -- Level

meaning:

In `Multilevel Inheritance`, Inheritance happens in levels.

### 📌 Structure

`Grandparent ➼ Parent ➼ Child`

### Why it is used?
* To represents heirarchy in real world,
* Step-by-step inheritance of feature.

### Real-Life Example

* Grandparent: Animal
* Parent: Mammal
* Child: Dog,Human

Dog is a Mammal, Mammal is a Animal

````python
class Animal: # Grand Parent
    def eat(self):
        print("Eating.......")

class Mammal(Animal):  # parent
    def walk(self):
        print("Walkng.....")
        
class Dog(Mammal):  # child   - Derived Child 
    def bark(self):
        print("Barking.....")

obj = Dog()
obj.eat()
obj.walk()
obj.bark()
````

````python
print(Dog.mro())
````

### 4. Hierarchical Inheritance

Meaning

In `Hierarchical Inheritance`,Multiple child classes inherits from a single parent class.


### 📌 Structure

```
            Parent1
           /   |   \
          /    |    \
         /     |     \
    child1   child2  child3

```


#### Why It is used?

* One parent has common functionality.
* Different children use an extend it different.

### Real-Life Example

* Parent: Employee
* Children:
    * Manager
    * Developer
    * Tester


#### All are employees but have different roles.

````python
class Employee: # parent
    def work(self):
        print("All Employees are working.....")

class Manager(Employee):
    def manage(self):
        print("Manager is managing work......")

class Developer(Employee):
    def code(self):
        print("Developer is coding....")
        
class Tester(Employee):
    def test(self):
        print("Tester is Testing......")

obj1 = Manager()
obj1.work()  # parent
obj1.manage()


obj2 = Developer()
obj2.work()  # parent
obj2.code()

obj3 = Tester()
obj3.work() # parent
obj3.test()
````

````python
print(Manager.mro())
````

````python
print(Developer.mro())
````

````python
print(Tester.mro())
````

### 5. Hybrid Inheritane

Meaning:

Hybrid inheritance is combination of two or more inheritance types.

It May Include:

* Multiple + Multilevel
* Hierarchical + Multiple

📌 Structure

```
            A
           / \
          B   C
           \ /
            D
```
Here:

 * B and C inherits A [Hierarchial]
 * D inherits from B and C [Multiple Inheritance]

### Why it is used?

* When a system requires complex relationship.
* Common in large frameword and enterprises.

Python Example

````python
class A:
    def showA(self):
        print("Class A")

class B(A):
    def showB(self):
        print("Class B")

class C(A):
    def showC(self):
        print("Class C")

class D(B,C):  # multiple
    def showD(self):
        print("Class D")

obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
````

````python
print(D.mro())
````

Key Points


### Benefits of Inheritance

* Supports polymorphism
* Imporve Modular design

### 1. Method Overloading(Conceptual in Python/Compile time Polymorphism): Same Method but different parameter


Python Does not support traditiional overloading, but we can achieve it using default parameter.

````python
class Calculator:
    def add(self,a=0,b=0,c=0):
        return a+b+c
calc = Calculator()
print(calc.add()) # No parameter
print(calc.add(10)) # one parameter
print(calc.add(10,20)) # two parameter
print(calc.add(10,20,30))  # three parameter
````

````python
### *args
class Calculator:
    def add(self,*args):
        return sum(args)
calc = Calculator()
print(calc.add()) # No parameter
print(calc.add(10)) # one parameter
print(calc.add(10,20)) # two parameter
print(calc.add(10,20,30))  # three parameter
print(calc.add(10,10,10,320,30,60,60,30,60,60,60,65,52))
````

KeyPoints:

* Same method name `add`
* Different number of argumets
* Behavior changes automatically.

````python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self._salary = salary  ## Protected

    # create method to access protected varaibles 
    def salaryDetails(self):
        return f"Current Salary is:{self._salary}"


emp = Employee("Ram", 45000)
print(emp.name)  # name is public because you can directly access outside the class.
emp.salaryDetails()  # salary is not accessible directly because it is protected you can access with the help of methods. 
# emp.salary
````

### 3. Private Members(`__` Double Underscore)


````python
class BankAccount:
    def __init__(self,name,salary):
        self.name = name  # public becuse here don't used access modifiers
        self.__salary = salary  # permently hidden

    ## display balance
    def show_balance(self):
        return f"Current Balance:{self.__salary}"


    # update balanace 

    def update_balance(self):
        amount = int(input("Enter any amount: "))
        if amount>=0:
            self.__salary = amount
            return f"Amount Credited Successfully:{amount}"

bank = BankAccount("Ram",45000)
print(bank.name)
print(bank.show_balance())  # Accessible with the help of method 
# print(bank.__salary)  # not accessess outside the class because it is private[permanently hidden]
        
````

📌 Structure

### How Abstraction

Python Provides: 

1. Abstract Base Classes(ABC)
2. Abstract method


### 1. Abstract Base Classe(Core Concept)

* Cannot be instanciated.
* Contains abstract method [Built-in]
* Acts as a blueprint

### 2. Abstract Method

* Has declaration only.
* No Implementation
* Must be implemented by child class

````python
from abc import ABC,abstractmethod
````

````python
class Vehicle(ABC):

    @abstractmethod
    def engine(self):
        pass

    @abstractmethod
    def start(self):
        print("Display Start")
        pass
# obj = Vehicle()
# obj.start()
````

````python
class Car(Vehicle):
    def automaticCar(self):
        print("My Car is automatic.")


    def engine(self):
        return f"This car has BS-IV engine"

    def start(self):
        return f"My Car is Started"

        
obj = Car()
obj.automaticCar()
obj.start()
obj.engine()
````

````python
class Shape:
    def area(self):
        pass


class Circle(Shape):

    def area(self):
        radius = int(input("Enter radius: "))
        return 3.14 * radius ** 2


class Rectangle(Shape):

    def area(self):
        length = int(input("Enter length: "))
        width = int(input("Enter width: "))
        return length * width


calc = Circle()
print("Area of Circle:", calc.area())

calc = Rectangle ()
print ("Area of Rectangle: ", calc.area())
````
