### 10.Exceptional Handling

#### Q.What is Exceptional Handling 
`Exceptional Handling` in Python is used to handle runtime Error so that the program does not crash suddenly and can continue execution smootly.

In Real-World software development, exception handle is mandatory becuase error can occurs anytime due to:

* ꪜWrong user input
* ꪜmissing file
* ꪜdata failure
* ꪜnetwork issue
* ꪜdivision by zero
* ꪜwrong data type
* ꪜinvalid operations

Python has mainly of two types of Error:

1. Compile Time error: When Your code is in Translation phase this errror occurred. ---> Developer/user

    * Occurs when python cannot understand your code.

2. RunTime Error: When you code is in execution time then this error occurred. ---> Machine can fix with the help of exceptional handling

   * Occurs while execution the program.
   * syntaxError

### 3. Why Exception Hanlding is Required?

#### ❌Without Exception Handling

  * ꪜProgram Crash
  * ꪜUser gets error message
  * ꪜsystem become unrealiable(Not Trust)
  * ꪜPrevent data loss

### With Exception Hanlding
  * ꪜProgram continue after getting error
  * ꪜWe can show user-friendly message.
  * ꪜBetter debugging and logging.
  * ꪜPrevent data loss

### Multiple Except Block

You can handle multiple Errors at a time

````python
x1 = "Hellow"
try:
    print(x1[0])
    print(10/0)
except ZeroDivisionError:
    print("ZeroDivisionError")
except NameError:
    print('NameError')
print("Done")
````

### 8. Single Except Handling Multiple Exceptions



````python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    div = x/y
except ZeroDivisionError,ValueError:
    print("ZeroDivisionError/ValueError")
else:
    print("Result:",div)
finally:
    print("Done")
    
````

#### try-except-else-finally

else block runs only if no exception occurs

````python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    div = x/y
except Exception as e:
    print(e)
else:
    print("Result:",div)
finally:
    print("Done")
    
````

Notes:

if exception ➺ except runs

if no exception ➺ else runs

### try-except-else-finally

### finally block runs (whether exception is happs or not)

````python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    div = x/y
except ZeroDivisionError,ValueError:
    print("ZeroDivisionError/ValueError")
else:
    print("Result:",div)
finally:  # file closing ---> File Handling 
    print("Program Done")
    
````
