### 10.Exceptional Handling

#### Q.What is Exceptional Handling 
`Exceptional Handling` in Python is used to handle runtime Error so that the program does not crash suddenly and can continue execution smootly.

In Real-World software development, exception handle is mandatory becuase error can occurs anytime due to:

* ꪜWrong user input
* ꪜmissing file
* ꪜdata failure
* ꪜnetwork issue
* ꪜdivision by zero
* ꪜwrong data type
* ꪜinvalid operations

```python
print("Hello"+3)
print("Hello world")
```

```python
try:
    print("Hello"+3)
except:
    print("Error")

print("Hello world")
```

### 2. Types of Errors in Python
Python has mainly of two types of Error:

1. Compile Time error: When Your code is in Translation phase this errror occurred. ---> Developer/user

    * Occurs when python cannot understand your code.

2. RunTime Error: When you code is in execution time then this error occurred. ---> Machine can fix with the help of exceptional handling

   * Occurs while execution the program.
   * syntaxError

```python
try:
    a = 10
    b = 0
    print(a/b)  # Runtime Error
except:
    print("Error: ZeroDivisionError")
print("Done")  # 
```

### 3. Why Exception Hanlding is Required?

#### ❌Without Exception Handling

  * ꪜProgram Crash
  * ꪜUser gets error message
  * ꪜsystem become unrealiable(Not Trust)
  * ꪜPrevent data loss

### With Exception Hanlding
  * ꪜProgram continue after getting error
  * ꪜWe can show user-friendly message.
  * ꪜBetter debugging and logging.
  * ꪜPrevent data loss
