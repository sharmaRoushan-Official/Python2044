### 13. Writing Files in Python

Writing Files means storing data permenently in a file from python program.

#### 📌 Used for:

* Saving user data
* Logs
* Reports
* Exploring results
* Backup files

-------

### File Writing Modes

|Mode|Meaning|
|----|-----|
|`w`|Write(Creates new/Overwrites file)|
|`a`|Append(adds data at the end)|
|`w+`|Write+Read|
|`a+`|Append+Read|
|`x`|Cretes new file if exists it throw error|


### Steps to write files

1. Open the file
2. Write data
3. Close the file

### 2. Writing Files using writelines()

```python
listFruits = ["Apple\n",'Mango\n',"Strawberry\n","Kiwi\n","Star Fruit\n","Grapes\n","Pineapple\n"]

file = open(r"C:\Users\User\Desktop\fruits.txt","w")

file.writelines(listFruits)

file.close()
```

```python
# Loop 

fileName = "C:\\Users\\User\\Desktop\\"
userName = input("Enter fileName")
completeFileName = fileName+userName+".txt"
print(completeFileName)
with open(completeFileName,'w') as file:
    lineCount = 0
    while True:
        lineCount+=1
        userData = input("Start Writing: ")
        if userData.capitalize() == "Stop":
            print("File Saved Successfully!!")
            break
        file.write(str(lineCount)+"."+userData)
        file.write("\n")
        
```

##### 7. Write+Read ("W+" Mode)

```python
with open("Sample.txt","w+") as file:
    file.write("Hello world")  # write the data
    file.seek(0)  # set pointer to 0 position
    print(file.read())  # read the data
    print(file.tell()) # where is the pointer right now
```

```python
# Store the path of the text file in a variable
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

# Create an empty list to store all extracted words
words = []

# Create an empty string to build words character by character
string = ""

# Open the file in read mode ('r')
with open(fileName, 'r') as f:

    # Read the complete content of the file
    content = f.read()

    # Iterate through each character of the file content
    for ch in content:

        # Check whether the current character is an alphabet
        if ch.isalpha():

            # Add the current alphabet to the string
            string += ch

        else:

            # Check whether the string contains a word
            if string:

                # Add the completed word to the words list
                words.append(string)

                # Make the string empty to start building the next word
                string = ""

    # Check whether the last word is still stored in the string
    # This handles files that do not end with a space or punctuation
    if string:

        # Add the last word to the words list
        words.append(string)

# Count and display the total number of words
print("Total Words:", len(words))

# Display the list of all extracted words
print("Words:", words)
```
