### 1. Positional Argument

````python
# right position
def student_details(name,age):
    return f"Student's name is {name}\nHis/Her Age is:{age}"


obj1 = student_details("Tanaz",23)
print(obj1)
````

````python
# Wrong Position
def student_details(name,age):
    return f"Student's name is {name}\nHis/Her Age is:{age}"


obj1 = student_details(100,"Tanaz")
print(obj1)
````

### 2. Keyword Argument

````python
def student_details(name,age):
    return f"Student's name is {name}\nHis/Her Age is:{age}"

obj1 = student_details(age=100,name="Tanaz")
print(obj1)
````

#### 7. Scope Variables

> 1. Global Variable
> 2. Local Variable

````python
age = 35 # global variable age 

def personAge():
    return age  # Accessing age inside function

personAge()
````

````python
print(age)
````

````python
# local scope/variable
def personAge():
    age1 = 35 # age1 is local varaible only access inside the function
    return age1  # Accessing age inside function

personAge()
````

````python
print(age1) # Not accessable outside the function
````

````python
# local scope/variable
age = 200
name = "Tanaz"
def personAge():
    age1 = 35 # age1 is local varaible only access inside the function
    # name = "apple"
    print(name)
    return age1  # Accessing age inside function

personAge()
````

````python
# global varaible 

def personName():
    global name1 # name1 become global varaible when you used global keyword before name1
    name1 = "Tanaz" # local variable 
    return name1

personName()
````

````python
print(name1)
````
