Q. What is for loop?

A `for loop` is used to iterate over a sequence(such as:list, tuple, set, dictionary,string,primitve data), and execute a block of code repeatedly.


syntax:

```
for variable in iterator:
    block of loop statement

```

Flow:
* Python takes the first item from the sequence
* Assign it in the loop variable.
* Execute the code block
* Move to the next item.
* Repeats untill all items are processed.

### Difference between for loop and while loop


|`for loop`|`while Loop`|
|----|---|
|Iterate over a sequence|Runs based on conditions|
|Known iteration|Unknown iteration|
|Simpler|More flexible|

------------

#### While Loop

A while loop repeatdly executes a block of code as long as condition is True.

* There is no range keyword
* You need to start and stop their vaule using conditions.
* You need to increment  thier value also
>> For Loop =  Known number of iterations
>> While Loop = Unknown number of iterations, depends on conditions.


### What is while loop?

A while loop check conditions before iterations.
