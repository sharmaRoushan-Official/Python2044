## 12. File Handling in Python

### What is File Handling?

File Handling allow python program to read data files and write files peranently.

📌 Why We read files:

* Stored data permently
* Read Logs,reports, CSV(Comma seperated value)
* Process large data
* Avoid manual input

#### Types of Files 
1. Text Files(.txt, .csv, .log)
2. Binary Files(.jpg,.pdf,.exe)


#### Steps to Read a file in Python


1. Open the file - RAM Memory
2. Read the file - CPU Process
3. close the file - Unautherized person can't access

syntax:

file  = open("fileName","r")




````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
print(fileName)

# step-1

file = open(fileName,'r')

# step - 2
content = file.read()
print(content)

## step - 3
file.close()
````

Notes:

|Mode|Meaning|
|---|----|
|`r`|Read(default)|
|`w`|Write|
|`a`|append|
|`r+`|Read and write|

### Method - 1: Read() ---> Entire file at once.
    

````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
print(fileName)

# step-1

file = open(fileName,'r')

# step - 2
content = file.read()
print(content)

## step - 3
file.close()
````

Notes:

* Read whole file at once.
* No suitable for very large file.

### Method 2: readline() -> Read Line by Line [One line at a time.]


````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
print(fileName)

# step-1

file = open(fileName,'r')

# step - 2
content = file.readline()
content2 = file.readline()
print(content)
print(content2)

## step - 3
file.close()
````

````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
print(fileName)

# step-1

file = open(fileName,'r')

# step - 2
content = file.readlines()
print(content)

## step - 3
file.close()
````

### Syntax for opening a file:

file_object = open("fileName",'mode')

Parameters:

----
|Parameter|Description|
|------|-------|
|`fileName`|Name of the file|
|`Mode`|Operations to perform|

------

3. Files Modes

----

|Mode|Description|
|----|-----|
|`r`|Read File(defualt)|
|`w`|Write file(Overwrite)|
|`a`|Append data|
|`x`|Create if exists it throw error|
|`wb`|Write binary|


### Reading File Using Loop:

Best method for large files.

````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
print(fileName)

file  = open(fileName,'r')

for line in file:
    print(line)
    
````

### File reading using `with` statement [Best Recommendation]

With automatically closes the file.


syntax:
```
with open("filename","mode") as file:

````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

with open(fileName,'r') as file:
    print(file.read())
````

````python
# fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

with open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r') as file:
    print(file.read())
````
