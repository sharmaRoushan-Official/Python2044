## 6. Dictionary in Python

-----
1. What is Dictionary in Python?

A `dictionary` is a collection of key:values pairs.

* Each key is unique
* Each key maps to a value.
* Dictionary are mutable(unhashable/changable)
* It is written using curly braces {}


```python
d1 = {}
print(d1,type(d1))
```

```python
s1 = set()
print(s1,type(s1))
```

```python
student = {"name":"Tanaz",'age':25}
print(student)
```

```python
student = {
    "name":"Tanaz",
    'age':25
}
print(student)
```

Note:

* name,age --> keys
* "Tanaz",25 ---> values

## 2. Why we use Dictionaries

Use dictionaries when data has a relationship between key and values

Example:

|key|value|
|---|----|
|"name"|"Tanaz"|
|'age'|25|


```python
employee = {
    'id':101,
    "name":"Amit",
    "Salary":50000
}
print(employee)
```

## 3. Chracteristics of Dictionary

Important points

1. Key must be unique

```python
data = {
    "name": "Tanaz",
    "name" : "apple"  # lastone 
}
print(data)
```

2. Keys must be immutable

allowed

* string
* number
* tuple

Not allowed: mutable

* list
* dictionary
* set


```python
data = {
    1:"Python",
    "name":"Arnav",
    ("Hello world",20):"Tuple key",
    # True: "Boolean" 
    False:"Boolean"
}
data
```

## 11. Looping Through Dictionary

```python
data1
```

```python
for key in data1:
    print(key)
```

```python
for key in data1.keys():
    print(key)
```

```python
for value in data1.values():
    print(value)
```

```python
for key,value in data1.items(): # : line terminator   ()
    print(key,":",value)  # indentation
```

#### 12. Nested Dictionary

```python
students = {
    "student1":{
        "name":"Arbaz",
        'age':17
    },
    "student2":{
        "name":"Hemant",
        "age":25
    },
    "student3":{
        "name":"Tanaz",
        "age":15
    }
}
print(students)
```
