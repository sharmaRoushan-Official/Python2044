### 13. Case Statement



````python
day = int(input("Enter week name: "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Satruday")
    case 7:
        print("Sunday")
    case _ :
        print("Invalid Week Day")
````

````python
food1 = ["Poha", "Upma", "Idli", "Meduwada"]

food2 = ["MishalPaw", "VadaPav", "Dosa", "Uttapa"]

food = input("Enter food item: ").lower()

# Convert list items to lowercase for comparison
if food in [item.lower() for item in food1]:
    price = 25
    print(f"{food.capitalize()} price is ₹{price}")

elif food in [item.lower() for item in food2]:
    price = 50
    print(f"{food.capitalize()} price is ₹{price}")

else:
    print("Food Not Available")
````

### 15. Write a Python program to check wheather a charater is a digit, upper latter, lower letter, or special chracter.



````python
value = input("Enter any value: ")

if ord(value)>=65 and ord(value)<=90:
    print(f"You entered upper case: {value}")
elif ord(value)>=97 and ord(value)<=122:
    print(f"You entered Smaller case:{value}")
elif ord(value)>=48 and ord(value)<=57:
    print(f"You entered digit: {value}")
elif ord(value)>=32 and ord(value)>=58 and ord(value)<=64:
    print(f"You entered symbol:{value}")
else:
    print("No chracter defined")
````
