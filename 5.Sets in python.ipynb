## 5. Sets in Python

-----

### 1. Introduction of sets

Logic: Mathematical calculations

A `set` in python is a collection of data type used to store multiple `unique` elements.

It is based on the mathematical conpcet of set.

### Key chracteristics

1. Unordered - No index value 
2. Unindexed
3. Mutable data type(can add/remove items)
4. Does not allow duplicate value
5. Elements must be immutable(hashable) 
----


```python
# How to create empty set 

s = {} # by default {} is dictionary data type
print(s,type(s))
```

```python

s = {1} # by default {} is dictionary data type
print(s,type(s))
```

```python
s2 = set()  # empty or convert into set.
print(s2,type(s2))
```

### 2. Creates sets

Method 1: using curly braces {}

```python
s = {1,2,3,4,5,6,7,8}
print(s,type(s))
```

## method 2: Using set() constructor

```python
list1 = [10,10,10,20,30,40,50,50]
print(list1)  # unhashable

s3 = set(list1) # list1 converted into set
print(s3,type(s3))
```

## 3. Duplicate Elements

set automatically remove duplicates



```python
s = {10,20,30,40,50,50,60,60,66,66,666,666}
print(s,type(s))  # sets hold only unique keys
```

### 4. Data type allowed in sets

Elements must be immutable(hashable)

Allowed:

* string
* tuple
* int
* float
* boolean
* frozenset

```python
s = {"hello world",(10,20,30),100,145.63,True,False, frozenset([10,10])}
print(s,type(s))
```

Not Allowed data type: Mutable(unhashable)

* list
* dictionary
* set

```python
set1 = {10,20,[10,20,30]}
print(set1,type(set1))
```

```python
set1 = {10,20,{"a":'apple'}}
print(set1,type(set1))
```

```python
set1 = {10,20,{10,10,20}}
print(set1,type(set1))
```

### 10. Set Operations(Most import point)

Python set supports mathematical operations 
1. union
2. intersection
3. difference
4. symmetric_difference



```python
A = {1,2,3}
B = {3,4,5}

print(A)
print(B)
```

## 1. Union - return unique value

1. Method - .union()
2. Symbol - | (pipe)

```python
print(A.union(B))
```

```python
print(A | B)
```

# 2. intersection

```python
## method - .intersection
### symbol - &
```

## 16. Built in Function in sets

|function|Description|
|----|----|
|`len()`| number of elements|
|`max()`| largest key in the set|
|`min()`| smallest key in the set|
|`sum()`| sum of the keys in the set|
|`sorted()`|Arranged in ascending order|

```python
s = {3,1,4}
print(s)
```

```python
print(len(s))
print(max(s))
print(min(s))
print(sum(s))
print(sorted(s))
```

## 17. Performance Advantage of Sets

Sets are used as hash table, so operations are very fast.



18. When to Use Sets

Use sets when:

* ✅ Remove duplicate
* ✅ Fast searching
* ✅ mathemtical Opetions: union, intersection, difference, symmetric_difference
* ✅ Membership testing

Example:

* Unique User ID
* Unique email address
* Tag system
* Data Filtering
