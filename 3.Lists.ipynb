* Including(Mixed data type)
* Lists are one of most used data types in python.

### 8. List Methods

---
|Method|Work|
|----|----|
|append(x)|add item at end|
|extend(iterable)|Add multiple items|
|insert(i,x)|i is index value and x is value|
|remove(n)| n is value|
|pop(n)| n is index value|
|pop()| remove last element|
|clear()| clear all values|
|del(n)|n is variable|
|index(x)| x is value return index value|
|count(x)| count occurrence of x value|
|reverse()| reverse the list|
|copy()| return shallow copy|

### Copy()

1. shallow copy(): reference copy
2. deep copy(): address copy()

1. shallow copy()


```python
x = [10,20,30]
print("Original List:",x)


y = x.copy()
print("Copied Value:",y)
```

```python
x.append(100)
x
```

```python
y
```

```python
id(x)
```

```python
id(y)
```

2. Deep Copy(): Address copy

```python
x = [10,20,30]
print("Original List:",x)


y = x
print("Copied Value:",y)
```

```python
id(x)
```

```python
id(y)
```

```python
x.append("apple")
x
```

```python
print(y)
```
