## 3. Lists in Python

* A `List` in Python is a mutable(Can be changes), Ordered, Indexed collections that can store multiple values.
* Including(Mixed data type)
* Lists are one of most used data types in python.


### 1. List Definition

---
Syntax:
`my_list = [1,2,3]`

```python
myList = []
print(myList,type(myList))
```

```python
data = [10,20,30]
print(data)
```

## 2. Feature of List

---
|Feature|Explanation|
|---|---|
|Ordered|Item maintain insertion order|
|Indexed|Each element has an index|
|Mutable|Can be modify after creation|
|Allows Duplicate|Same values can repeat|
|Mixed types allowed| int, float, str, bool, object-[string, list, tuple, set, dictionary]

```python
x = [10,25.85,"python",True,False, [10,20],(10,14),{10,12},{"a":"apple"}]
print(x)
```

### Index - Positive

```python
print(x[0])
print(x[1])
print(x[2])
print(x[3])
print(x[4])
print(x[5])
print(x[6])
print(x[7])
print(x[8])
```

### Index Negative

```python
print(x[-1])
print(x[-2])
print(x[-3])
print(x[-4])
print(x[-5])
print(x[-6])
print(x[-7])
print(x[-8])
print(x[-9])
```

## type()

```python
print(x[0])
print(type(x[0]))

print(x[1])
print(type(x[1]))

print(x[2])
print(type(x[2]))
```

```python
x = [10,25.85,"python",True,False, [10,20],(10,14),{10,12},{"a":"apple"}]
print(x)
```

## 4. Slicing of List

`var[start:end]`

```python
x[0:5]
```

```python
x[7:9]
```

```python
x[3:7]
```

```python
x = [10,[10,20,30],[[10,20,50],True,False],[14,17,89],("Python",50),30,60,78,90]

x[2][0][0]
```
